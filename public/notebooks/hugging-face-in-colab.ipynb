## 1. Check the machine

Colab already has PyTorch and 🤗 Transformers installed. This cell shows the versions and whether a GPU is available.

```python
import torch, transformers
print("PyTorch", torch.__version__, "| Transformers", transformers.__version__)
print("GPU available:", torch.cuda.is_available())
device = 0 if torch.cuda.is_available() else -1
```

## 2. Your first pipeline

`pipeline()` downloads a model from the Hub the first time, then runs it. This one is DistilBERT fine-tuned for English sentiment (Apache-2.0) — the same model as the in-browser demo in the lab.

```python
from transformers import pipeline

clf = pipeline("sentiment-analysis",
               model="distilbert/distilbert-base-uncased-finetuned-sst-2-english",
               device=device)
clf(["The delivery was fast and the agent was very helpful!",
     "The app keeps crashing and nobody answers my calls."])
```

## 3. Swahili → English translation

NLLB-200 from Meta translates between 200 languages, including Swahili (`swh_Latn`).

⚠️ **Check the licence:** NLLB is released under **CC BY-NC 4.0** — fine for learning and research, *not* for commercial products. Always read the model card before you build on a model.

```python
translator = pipeline("translation",
                      model="facebook/nllb-200-distilled-600M",
                      src_lang="swh_Latn", tgt_lang="eng_Latn",
                      device=device)

reviews = [
    "Huduma nzuri sana, asante!",
    "Mbegu hazikumea na bei ni ghali.",
    "Programu inafanya kazi vizuri.",
]
translations = [t["translation_text"] for t in translator(reviews)]
translations
```

## 4. Chain two models

Translate Swahili reviews to English, then score their sentiment. A quick way to get *some* Swahili sentiment — with the errors of both models combined.

```python
for sw, en, s in zip(reviews, translations, clf(translations)):
    print(f"{sw}\n  → {en}\n  → {s['label']} ({s['score']:.2f})\n")
```
